```python
## Load packages and parameters
import os
import scanpy as sc
import anndata as ad
from anndata import AnnData
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.colors as mcolors
from matplotlib.patches import Patch
from scipy import sparse
import gc
from joblib import Parallel, delayed
from sklearn.neighbors import NearestNeighbors
from ALLCools.mcds import MCDS
from scipy.stats import mode, pearsonr
import pybedtools
import re
import xarray as xr
pybedtools.helpers.set_bedtools_path("/jp_envs/envs/seeksoulmethyl/bin/")

## Load MethImpute
import sys
sys.path.append("/home/lishengdi/workspace/project/lishengdi/scPRIMEseq_pynotebooks/")
#import MethImpute
import MethImpute as methimp
```

```python
## Step 1. Generate mC fraction matrix from ALLCools output

DEBUG_IDS = {"RS67"}   # or None to run all
#DEBUG_IDS = None

sample_meta_path = "../input_metadata/sample_data.txt"
var_dim="chrom20k"
#var_dim="geneslop2k"
#var_dim="promoter"
mcds_output = f"tmp_dataset/mcds_{var_dim}/"

meta = pd.read_csv(sample_meta_path, sep="\t")
required_cols = {"sample_id", "allcool_gen_matrix","filtered_barcode"}

missing = required_cols - set(meta.columns)
if missing:
    raise ValueError(f"Missing required columns: {missing}")

black_list_path = "../input_metadata/hg38-blacklist.v2.bed"
black_list_fraction = 0.2
exclude_chrom = ['chrM', 'chrY']


### calcualte mcFraction and store the per patient mcds file
if DEBUG_IDS:
    meta = meta[meta["sample_id"].isin(DEBUG_IDS)]
for _, row in meta.iterrows():
    sample_id = row["sample_id"]
    mcds_path = row["allcool_gen_matrix"]
    mcds = MCDS.open(str(mcds_path), var_dim=var_dim)
    
    mcds = mcds.remove_black_list_region(
        black_list_path=black_list_path,
        f=black_list_fraction  # Features having overlap > f with any black list region will be removed.
    )
    mcds = mcds.remove_chromosome(exclude_chrom)

    old_cells = mcds.coords["cell"].values
    new_cells = np.array([f"{sample_id}_{c}" for c in old_cells])
    mcds = mcds.assign_coords(cell=new_cells)
    mcds.add_mc_frac(normalize_per_cell=False, clip_norm_value=None, da_suffix = 'true_frac')
    mcds_da_true_frac = mcds[[f'{var_dim}_da_true_frac']]
    mcds_da_true_frac[f'{var_dim}_da_true_frac'] = mcds_da_true_frac[f'{var_dim}_da_true_frac'].astype('float32')
    mcds_da_true_frac.write_dataset(output_path=f"{mcds_output}/{sample_id}_{var_dim}_da_true_frac",mode="w")

    mcds.add_mc_frac(normalize_per_cell=True, clip_norm_value=10)
    mcds_da_frac = mcds[[f'{var_dim}_da_frac']]
    mcds_da_frac[f'{var_dim}_da_frac'] = mcds_da_frac[f'{var_dim}_da_frac'].astype('float32')
    mcds_da_frac.write_dataset(output_path=f"{mcds_output}/{sample_id}_{var_dim}_da_frac",mode="w")
    
```

```python

## Step 2. Load RNA matrix and perform imputation of methylation matrix

DEBUG_IDS = {"RS67"}   # or None to run all
#DEBUG_IDS = None

cell_meta = pd.read_csv("../input_metadata/sample_data.txt", sep="\t")
var_dim = "promoter" # or change to "geneslop2k"
output_path = f"tmp_dataset/h5ad_{var_dim}/"
os.makedirs(output_path, exist_ok=True)

### generate impute matrix for all sample for geneslop2k, use k = 60, min_cell = 5
### generate impute matrix for all sample for promoter, use k = 100, min_cell = 2
knn_k = 100
min_cell = 2

if DEBUG_IDS:
    cell_meta = cell_meta[cell_meta["sample_id"].isin(DEBUG_IDS)]
for _, row in cell_meta.iterrows():
    p_id = row["sample_id"]
    rna_path = row["RNA_matrix"]
    mcds_path = f"tmp_dataset/mcds_{var_dim}/{p_id}_{var_dim}_da_true_frac"
    #out_path = f"{output_path}/{p_id}_geneslop2k_imputed.h5ad"
    out_path = f"{output_path}/{p_id}_{var_dim}_imputed.h5ad"
    #out_path = f"{output_path}/{p_id}_chrom20k_imputed.h5ad"
    if os.path.exists(out_path):
        print(f"{p_id} already processed, skipping.")
        continue
    rna_adata = methimp.load_10x_rna(rna_path)
    rna_adata = methimp.preprocess_rna(rna_adata)
    #### adjust cell_id
    rna_adata.obs_names = p_id + "_" + rna_adata.obs_names
    rna_adata.obs_names

    ### 2. load and transform methy matrix (-1:missing; 1:hypo; 0:hyper)
    methy_adata = methimp.load_mcds_sample_wo_categorize(mcds_path,data_dim=f"{var_dim}_da_true_frac",var_dim=var_dim)
    
    ### 3. intersecting RNA and methy cell_ids
    common_cells = rna_adata.obs_names.intersection(methy_adata.obs_names)
    rna_adata = rna_adata[common_cells, :].copy()
    methy_adata = methy_adata[common_cells, :].copy()
    rna_pca = rna_adata.obsm['X_pca']
    knn_idx, knn_dist = methimp.build_knn(rna_pca, k=knn_k)
    
    meth_matrix = methy_adata.X
    if sparse.issparse(meth_matrix):
        meth_matrix = meth_matrix.toarray()
    ### the most frequent value in each cell is the missing data, set as -1
    row_mode = mode(meth_matrix, axis=1, keepdims=True).mode  # shape (n_cells, 1)
    meth_matrix = np.where(meth_matrix == row_mode, -1, meth_matrix)
    imputed_meth = methimp.knn_impute_weighted_rate_sparse(meth_matrix,knn_idx,knn_dist,n_jobs=24,min_cell = min_cell)
    methy_adata.X = imputed_meth
    
    methy_adata.write_h5ad(out_path)
    print(f"Finished {p_id}")
    
```

```python
## Step 2b (chrom20K, binarized). Load RNA matrix and perform imputation of methylation matrix

DEBUG_IDS = {"RS67"}   # or None to run all
#DEBUG_IDS = None

cell_meta = pd.read_csv("../input_metadata/sample_data.txt", sep="\t")
var_dim = "chrom20k"
output_path = f"tmp_dataset/methy_{var_dim}/"
os.makedirs(output_path, exist_ok=True)

knn_k = 20
min_cell = 2

if DEBUG_IDS:
    cell_meta = cell_meta[cell_meta["sample_id"].isin(DEBUG_IDS)]
for _, row in cell_meta.iterrows():
    p_id = row["sample_id"]
    rna_path = row["RNA_matrix"]
    mcds_path = f"tmp_dataset/mcds_{var_dim}/{p_id}_{var_dim}_da_frac"
    #out_path = f"{output_path}/{p_id}_geneslop2k_imputed.h5ad"
    out_path = f"{output_path}/{p_id}_{var_dim}_imputed.h5ad"
    #out_path = f"{output_path}/{p_id}_chrom20k_imputed.h5ad"
    if os.path.exists(out_path):
        print(f"{p_id} already processed, skipping.")
        continue
    rna_adata = methimp.load_10x_rna(rna_path)
    rna_adata = methimp.preprocess_rna(rna_adata)
    #### adjust cell_id
    rna_adata.obs_names = p_id + "_" + rna_adata.obs_names
    rna_adata.obs_names

    ### 2. load and transform methy matrix (-1:missing; 1:hypo; 0:hyper)
    methy_adata = methimp.load_mcds_sample(mcds_path,data_dim=f"{var_dim}_da_frac",var_dim=var_dim)
    
    ### 3. intersecting RNA and methy cell_ids
    common_cells = rna_adata.obs_names.intersection(methy_adata.obs_names)
    rna_adata = rna_adata[common_cells, :].copy()
    methy_adata = methy_adata[common_cells, :].copy()
    rna_pca = rna_adata.obsm['X_pca']
    knn_idx, knn_dist = methimp.build_knn(rna_pca, k=knn_k)
    
    meth_matrix = methy_adata.X
    if sparse.issparse(meth_matrix):
        meth_matrix = meth_matrix.toarray()
    imputed_meth = methimp.knn_impute_weighted_parallel(meth_matrix,knn_idx,knn_dist,n_jobs=24,min_cell = min_cell)
    methy_adata.layers['raw'] = methy_adata.X ### annotate this out for saving disk space
    methy_adata.X = sparse.csr_matrix(imputed_meth)
    methy_adata.write_h5ad(out_path)
    print(f"Finished {p_id}")
    
```

```python
## Visualize before and after imputation for chr8
cell_meta = pd.read_csv("../input_metadata/cell_meta.tsv", sep="\t")
cell_meta = cell_meta.set_index('cell_id')  # index = cell IDs
methy_adata = sc.read_h5ad("tmp_dataset/methy_chrom20k/RS67_chrom20k_imputed.h5ad")
cell_meta["is_LE"] = (cell_meta["epi_cl"] == "LE")
pb_matrix = methy_adata.X
pb_matrix_raw = methy_adata.layers['raw']
if sparse.issparse(pb_matrix):
    pb_matrix = pb_matrix.toarray()
    pb_matrix_raw = pb_matrix_raw.toarray()
chr8_mask = methy_adata.var_names.str.contains("chr8")
pb_chr8_matrix = pb_matrix[:, chr8_mask]
pb_chr8_matrix_raw = pb_matrix_raw[:, chr8_mask]
matrix_cells = methy_adata.obs_names  # all cells in pb_chr8_matrix
common_cells = [c for c in matrix_cells if c in cell_meta.index]
pb_chr8_matrix = pb_chr8_matrix[[methy_adata.obs_names.get_loc(c) for c in common_cells], :]
pb_chr8_matrix_raw = pb_chr8_matrix_raw[[methy_adata.obs_names.get_loc(c) for c in common_cells], :]
matrix_cells_filtered = common_cells  # already aligned
cell_meta_sub = cell_meta.loc[matrix_cells_filtered]
cell_meta_sub['ploidy_filled'] = cell_meta_sub['ploidy'].fillna('NA')
cell_meta_sorted = cell_meta_sub.sort_values(
    by=['cell_type', 'ploidy_filled',"is_LE"],  
    ascending=[True, True, True]      
)
sort_idx = [cell_meta_sub.index.get_loc(c) for c in cell_meta_sorted.index]
pb_chr8_matrix_sorted = pb_chr8_matrix[sort_idx, :]
pb_chr8_matrix_raw_sorted = pb_chr8_matrix_raw[sort_idx, :]
unique_ct = cell_meta_sorted["cell_type"].unique()
palette = [
  "#1F78B4",  # blue
  "#33A02C",  # green
  "#E31A1C",  # red
  "#FF7F00",  # orange
  "#6A3D9A",  # purple
  "#B15928",  # brown
  "#4DBBD5",  # cyan
  "#00A087",  # teal
  "#F39B7F",  # salmon
  "#666666"   # gray
]
cell_type_to_color = dict(zip(unique_ct, palette))

cell_colors = cell_meta_sorted["cell_type"].map(cell_type_to_color)
rgb_colors = np.array([mcolors.to_rgb(c) for c in cell_colors])
rgb_colors_2d = rgb_colors[np.newaxis, :, :]
ploidy_cat = cell_meta_sorted['ploidy'].astype(str)  # 确保是字符串类型
unique_ploidy = ploidy_cat.unique()
ploidy_palette = sns.color_palette("Set2", len(unique_ploidy))
ploidy_to_color = dict(zip(unique_ploidy, ploidy_palette))
ploidy_colors = np.array([mcolors.to_rgb(ploidy_to_color[p]) for p in ploidy_cat])
ploidy_colors_2d = ploidy_colors[np.newaxis, :, :]  
le_cat = cell_meta_sorted['is_LE'].astype(str)  
unique_le = le_cat.unique()
le_palette = sns.color_palette("Set3", len(unique_le))
le_to_color = dict(zip(unique_le, le_palette))
le_colors = np.array([mcolors.to_rgb(le_to_color[p]) for p in le_cat])
le_colors_2d = le_colors[np.newaxis, :, :]  
annotation_colors_2d = np.vstack([rgb_colors_2d, ploidy_colors_2d, le_colors_2d])  # shape (2, N, 3)
annotation_colors_2d = np.repeat(annotation_colors_2d, 10, axis=0)  # stretch for visibility

cmap = mcolors.LinearSegmentedColormap.from_list(
    "yellow_white_blue", 
    ["white", "#2166AC", "#FFD700"] 
)

norm = mcolors.TwoSlopeNorm(vmin=-1,
                            vcenter=0, 
                            vmax=1)

fig = plt.figure(figsize=(15, 8))
gs = fig.add_gridspec(3, 1, height_ratios=[0.1, 1, 1], hspace=0.05)
ax0 = fig.add_subplot(gs[0])
ax0.imshow(annotation_colors_2d, aspect='auto')
ax0.set_xticks([])
ax0.set_yticks([0,1,2])
ax0.set_yticklabels(['Cell Type','Ploidy','is_LE'], rotation=0)
ax0.set_ylabel("")
ax0.axis('off')  # remove ticks
ax1 = fig.add_subplot(gs[1], sharex=ax0)
sns.heatmap(pb_chr8_matrix_sorted.T,
            cmap=cmap,
            norm=norm,
            cbar=False,
            xticklabels=False,
            yticklabels=False,
            rasterized=True,
            ax=ax1)
ax1.set_ylabel("Imputed Methylation")

ax2 = fig.add_subplot(gs[2], sharex=ax0)
sns.heatmap(pb_chr8_matrix_raw_sorted.T,
            cmap=cmap,
            norm=norm,
            cbar=False,
            xticklabels=False,
            yticklabels=False,
            rasterized=True,
            ax=ax2)
ax2.set_xlabel("Cells (ordered by cell type)")
ax2.set_ylabel("Raw Methylation")

legend_elements = [Patch(facecolor=col, label=ct) for ct, col in cell_type_to_color.items()]
ax2.legend(handles=legend_elements, title="Cell Type", bbox_to_anchor=(1.05, 1), loc='upper left')

plt.rcParams['pdf.fonttype'] = 42
#plt.savefig("figures/RS65_chr8_impute_chrom20k.pdf", bbox_inches="tight")
plt.show()
```

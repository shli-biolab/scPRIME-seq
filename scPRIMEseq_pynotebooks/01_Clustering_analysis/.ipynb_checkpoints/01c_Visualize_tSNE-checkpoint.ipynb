```r
## Annotate cell type
metadata <- read_tsv("tmp_dataset/rna_consensus_clustering.tsv")
metadata <- metadata %>%
  mutate(sample_id = str_replace(cell_id, "_[^_]+$", ""))

cluster_to_celltype <- c(
  "0" = "Epithelial",
  "1" = "Epithelial",
  "2" = "Fibroblast",
  "3" = "Endothelial",
  "4" = "Epithelial",
  "5" = "T",
  "6" = "T",
  "7" = "Myeloid",
  "8" = "Smooth muscle",
  "9" = "Endothelial",
  "10" = "Smooth muscle",
  "11" = "Mast",
  "12" = "B",
  "13" = "Epithelial",
  "14" = "Plasma",
  "15" = "Epithelial",
  "16" = "Epithelial",
  "17" = "Endothelial",
  "18" = "Endothelial",
  "19" = "Epithelial",
  "20" = "Schwann & other",
  "21" = "Epithelial",
  "22" = "Epithelial",
  "23" = "Fibroblast"
)

metadata$cell_type <- cluster_to_celltype[as.character(metadata$rna_cluster)]

#filter out small clusters with <1k cells
metadata <- metadata %>%
  filter(rna_consensus_cluster < 24)
```

```r
## Load packages
library(ggplot2)
library(gridExtra)
library(readr)
library(dplyr)
library(stringr)
library(ggrastr)
```

```r
## Annotate cell type
metadata <- read_tsv("tmp_dataset/rna_consensus_clustering.tsv")
metadata <- metadata %>%
  mutate(sample_id = str_replace(cell_id, "_[^_]+$", ""))

cluster_to_celltype <- c(
  "0" = "Epithelial",
  "1" = "Epithelial",
  "2" = "Fibroblast",
  "3" = "Endothelial",
  "4" = "Epithelial",
  "5" = "T",
  "6" = "T",
  "7" = "Myeloid",
  "8" = "Smooth muscle",
  "9" = "Endothelial",
  "10" = "Smooth muscle",
  "11" = "Mast",
  "12" = "B",
  "13" = "Epithelial",
  "14" = "Plasma",
  "15" = "Epithelial",
  "16" = "Epithelial",
  "17" = "Endothelial",
  "18" = "Endothelial",
  "19" = "Epithelial",
  "20" = "Schwann & other",
  "21" = "Epithelial",
  "22" = "Epithelial",
  "23" = "Fibroblast"
)

metadata$cell_type <- cluster_to_celltype[as.character(metadata$rna_cluster)]

#filter out small clusters with <1k cells
metadata <- metadata %>%
  filter(rna_consensus_cluster < 24)
```

```r
unique_celltypes <- sort(unique(metadata$cell_type))
colors <- c(
  "#1F78B4",  # blue
  "#33A02C",  # green
  "#E31A1C",  # red
  "#FF7F00",  # orange
  "#6A3D9A",  # purple
  "#B15928",  # brown
  "#4DBBD5",  # cyan
  "#00A087",  # teal
  "#F39B7F",  # salmon
  "#666666"   # gray
)

colors <- colors[1:length(unique_celltypes)]
names(colors) <- unique_celltypes

p <- ggplot(metadata, aes(x = TSNE1, y = TSNE2, color = cell_type)) +
  geom_point_rast(size = 0.004, alpha = 0.5) +  # rasterized points
  scale_color_manual(values = colors) +
  theme_minimal(base_size = 14) +
  theme(
    panel.grid = element_blank(),
    axis.text = element_text(size = 12),
    axis.title = element_text(face = "bold"),
    legend.title = element_text(face = "bold", size = 12),
    legend.text = element_text(size = 10)
  ) +
  labs(
    x = "tSNE 1",
    y = "tSNE 2",
    color = "Cell Type",
    title = "458484 cells"
  )

ggsave("figures/tsne_celltype_consensus_cluster.pdf", p, width = 12, height = 10)
```
